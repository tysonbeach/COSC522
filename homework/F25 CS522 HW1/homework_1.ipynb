```python
import csv
import math
from pathlib import Path

data_path = Path(
    "/Users/tyson/COSC522/homework/F25 CS522 HW1/data.txt"
)

with data_path.open(newline="") as file:
    reader = csv.reader(file)
    headers = next(reader)
    rows = [
        [int(value) for value in row]
        for row in reader
        if row
    ]


def entropy(rows):
    if not rows:
        return 0.0

    total = len(rows)
    result = 0.0

    for label in (0, 1):
        count = sum(row[-1] == label for row in rows)
        if count:
            probability = count / total
            result -= probability * math.log2(probability)

    return result


def information_gain(rows, attribute):
    if not rows:
        return 0.0

    conditional_entropy = 0.0

    for value in (0, 1):
        subset = [row for row in rows if row[attribute] == value]
        conditional_entropy += (
            len(subset) / len(rows)
        ) * entropy(subset)

    return entropy(rows) - conditional_entropy


gains = {
    attribute: information_gain(rows, attribute)
    for attribute in range(len(headers) - 1)
}

print(f"Label entropy: {entropy(rows):.4f}\n")

for attribute, gain in gains.items():
    print(f"{headers[attribute]}: {gain:.4f}")

best_attribute = max(gains, key=gains.get)

print(f"\nNode 1: {headers[best_attribute]}")
print(f"Information gain: {gains[best_attribute]:.4f}")
```

```python
# Section 1, Question 1(ii)-(iii): choose the splits at Nodes 2 and 3.
# Run the Node 1 cell first to define the data and helper functions.
remaining_attributes = [
    attribute
    for attribute in gains
    if attribute != best_attribute
]

for value, node_number in [(0, 2), (1, 3)]:
    subset = [
        row for row in rows
        if row[best_attribute] == value
    ]

    node_gains = {
        attribute: information_gain(subset, attribute)
        for attribute in remaining_attributes
    }

    print(f"\nNode {node_number}: {headers[best_attribute]} = {value}")
    print(f"Rows: {len(subset)}")
    print(f"Label entropy: {entropy(subset):.4f}")

    for attribute, gain in node_gains.items():
        print(f"  {headers[attribute]}: {gain:.4f}")

    chosen = max(node_gains, key=node_gains.get)
    print(f"Selected attribute: {headers[chosen]}")
    print(f"Information gain: {node_gains[chosen]:.4f}")
```
